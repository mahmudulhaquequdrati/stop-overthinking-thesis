## Block 5–6: loop and if
**Problem:** only correct answers count for "room to shorten". **Why:** lesson 05.
**In:** 4 lengths and 4 pass/fail results. **Out:** the correct lengths. **Why this way:** a loop checks each try once.

```python
passed = [True, True, False, True]
correct_lengths = []
for length, ok in zip(lengths, passed):
    if ok:
        correct_lengths.append(length)
    else:
        print("skipping a failed try of", length, "tokens")
print("correct lengths:", correct_lengths)
```

## Block 7: a function
**Problem:** we compute "room to shorten" for many problems. **Why:** write the recipe once, reuse it.
**In:** a list of correct lengths. **Out:** shortest ÷ average. **Why this way:** a function avoids copy-paste mistakes.

```python
def room_to_shorten(correct_lengths):
    average = sum(correct_lengths) / len(correct_lengths)
    return min(correct_lengths) / average

ratio = room_to_shorten(correct_lengths)
print("room to shorten:", round(ratio, 2))
if ratio <= 0.75:
    print("room to shorten ✅ (our rule: at most 0.75)")
else:
    print("little room to shorten")
```

## Try: the selection rule from lesson 05
**Problem:** which answer do we keep for training? **Why:** shortest correct, but not below half the median.
**In:** correct lengths. **Out:** the kept length. **Why this way:** same rule as DECISIONS #14, in a few lines.

```python
def keep_for_training(correct_lengths):
    s = sorted(correct_lengths)
    middle = len(s) // 2
    median = s[middle] if len(s) % 2 == 1 else (s[middle - 1] + s[middle]) / 2
    allowed = [x for x in s if x >= median / 2]
    return min(allowed), median

kept, median = keep_for_training([3000, 600, 2400])   # problem B from lesson 05
print("median:", median, "| half:", median / 2, "| kept:", kept)
```
